# 02 — Independent baselines and linear probes
Batch 64; seeds 14/27/42; 20 baseline epochs. Explicit execution is required. Record the printed RUN path for the next notebooks. Failed attempts are not silently resumed.

```python
from pathlib import Path
import json, subprocess, sys
PROJECT = Path('/kaggle/working/rq71') if Path('/kaggle').exists() else Path('/content/rq71')
DATA = Path('/kaggle/working/rq71_data') if Path('/kaggle').exists() else Path('/content/rq71_data')
SPLITS = PROJECT / 'splits'
RESULTS = Path('/kaggle/working/rq71_results') if Path('/kaggle').exists() else Path('/content/rq71_results')
assert (PROJECT / 'run_three_seeds.py').is_file(), 'Set PROJECT to the attached reviewed source'
```

```python
from datetime import datetime,timezone
RUN=RESULTS/datetime.now(timezone.utc).strftime('staged_%Y%m%dT%H%M%S_%fZ')
SEEDS=[14,27,42]
EXECUTE=False
if EXECUTE:
    RUN.mkdir(parents=True,exist_ok=False)
    for seed in SEEDS:
        cfg=json.loads((PROJECT/'configs/full.json').read_text())
        cfg.update(seed=seed,data_dir=str(DATA),manifest_dir=str(SPLITS),output_dir=str(RUN/f'seed_{seed}'),quarantine_cross_split_duplicates=False)
        path=RUN/f'config_seed{seed}.json'
        path.write_text(json.dumps(cfg,indent=2))
        subprocess.run([sys.executable,'-u',str(PROJECT/'run_stage.py'),'baselines','--config',str(path)],check=True)
    print('Save this RUN path:',RUN)
else:
    print('Execution disabled')
```
